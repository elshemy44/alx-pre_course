### Task 7.3: List Slicing

5. Reverse the entire list using slicing


````python
# Write your code here
letters = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h']

# 1. First three letters
print("First three letters:", letters[:3])

# 2. Last three letters
print("Last three letters:", letters[-3:])

# 3. Middle four letters (c, d, e, f)
print("Middle four letters:", letters[2:6])

# 4. Every second letter
print("Every second letter:", letters[::2])

# 5. Reverse the entire list using slicing
print("Reversed list:", letters[::-1])
````

## Section 9: Sets

### Task 9.1: Set Operations
Create two sets:
- `set_a = {1, 2, 3, 4, 5}`
- `set_b = {4, 5, 6, 7, 8}`

Calculate and print:
1. Union of both sets
2. Intersection of both sets
3. Difference (elements in set_a but not in set_b)
4. Symmetric difference (elements in either set but not both)


````python
# Write your code here
set_a = {1, 2, 3, 4, 5}
set_b = {4, 5, 6, 7, 8}
print("Union:", set_a | set_b)  # or set_a.union(set_b)

# 2. Intersection
print("Intersection:", set_a & set_b)  # or set_a.intersection(set_b)

# 3. Difference (elements in set_a but not in set_b)
print("Difference (set_a - set_b):", set_a - set_b)  # or set_a.difference(set_b)

# 4. Symmetric difference (elements in either set but not both)
print("Symmetric difference:", set_a ^ set_b)  # or set_a.symmetric_difference(set_b)
````

### Challenge 1: Grade Calculator
Create a program that:
1. Asks the user for 5 test scores (use input and convert to integers)
2. Stores them in a list
3. Calculates and prints:
   - Average score
   - Highest score
   - Lowest score
   - Number of scores above 70


````python
# Write your code here
# Step 1: Ask the user for 5 test scores
scores = []
for i in range(1, 6):
    score = int(input(f"Enter score {i}: "))
    scores.append(score)

# Step 2: Calculate average, highest, lowest, and count above 70
average = sum(scores) / len(scores)
highest = max(scores)
lowest = min(scores)
above_70 = sum(1 for s in scores if s > 70)

# Step 3: Print the results
print("\n--- Grade Summary ---")
print("Scores:", scores)
print("Average score:", average)
print("Highest score:", highest)
print("Lowest score:", lowest)
print("Number of scores above 70:", above_70)
````

### Challenge 2: Shopping Cart
Create a shopping cart system using a dictionary where:
- Keys are item names
- Values are dictionaries containing "price" and "quantity"

Example:
```python
cart = {
    "Apple": {"price": 2.5, "quantity": 3},
    "Bread": {"price": 3.0, "quantity": 1}
}
```

Calculate and print:
1. Total cost for each item (price × quantity)
2. Total cost of the entire cart
3. Number of items in the cart


````python
# Write your code here
# Step 1: Create the shopping cart dictionary
cart = {
    "Apple": {"price": 2.5, "quantity": 3},
    "Bread": {"price": 3.0, "quantity": 1},
    "Milk": {"price": 1.8, "quantity": 2},
    "Eggs": {"price": 0.5, "quantity": 12}
}

# Step 2: Calculate total cost per item and total cost of the cart
total_cart_cost = 0

print("--- Item Costs ---")
for item, info in cart.items():
    total_item = info["price"] * info["quantity"]
    total_cart_cost += total_item
    print(f"{item}: ${total_item:.2f}")

# Step 3: Print total cost of the cart
print("\nTotal cost of the cart: $", round(total_cart_cost, 2))

# Step 4: Print number of items in the cart
print("Number of items in the cart:", len(cart))
````

### Challenge 3: Word Counter
Ask the user to enter a sentence. Then:
1. Split the sentence into words (use `.split()`)
2. Count how many words there are
3. Find the longest word
4. Create a dictionary showing how many times each word appears

**Hint:** Convert everything to lowercase for accurate counting


````python
# Write your code here
# Step 1: Ask the user for a sentence
sentence = input("Enter a sentence: ").lower()  # convert to lowercase

# Step 2: Split the sentence into words
words = sentence.split()

# Step 3: Count the number of words
word_count = len(words)

# Step 4: Find the longest word
longest_word = max(words, key=len)

# Step 5: Count the frequency of each word
word_freq = {}
for word in words:
    word_freq[word] = word_freq.get(word, 0) + 1

# Step 6: Print results
print("\n--- Word Analysis ---")
print("Words:", words)
print("Total number of words:", word_count)
print("Longest word:", longest_word)
print("Word frequencies:", word_freq)
````
